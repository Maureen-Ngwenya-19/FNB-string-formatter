Write a Python script called string_formatter.py 

that takes a user’s first name, last name, and a short bio message as input, then applies

multiple string transformations to produce a formatted user profile output. This simulates how a 

real app backend processes user-submitted text.

```python
# Collect first name, last name, and bio message using input()

first = input("Enter your first name")
last = input("Enter your last name")
message = input ("Enter your bio message")
print(f"First Name: {first}\nLast Name: {last}\nBio Message: {message}")
```

```python
# Create a username by combining first initial + last name in lowercase (e.g. ‘tdlamini’)

username = f"{first[0]}{last}"
print(f"Your username is :{username.lower()}")
```

```python
# Display the full name in Title Case using .title()

first.title()
```

```python
# Strip leading/trailing whitespace from the bio before displaying it using .strip()

message.strip()
```

```python
# Count and display the number of characters in the bio using len()

print(f"Number of characters in the bio : {len(message)}")
```

```python
# Replace any occurrence of ‘I am’ in the bio with ‘I’m’ using .replace()
message = message.replace("I am", "I'm")

print(f"Updated Bio: {message}")
```
